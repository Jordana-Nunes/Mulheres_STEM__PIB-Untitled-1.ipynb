```python
# Participação das Mulheres na STEM vs Crescimento Econômico
# Data: World Bank API

import requests
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

sns.set(style="whitegrid")

# Função - baixar dados da World Bank
def download_worldbank(indicator):

    url = f"https://api.worldbank.org/v2/country/all/indicator/{indicator}?format=json&per_page=20000"

    r = requests.get(url)
    data = r.json()

    if len(data) < 2:
        print(f"Nenhum dado para {indicator}")
        return pd.DataFrame()

    df = pd.DataFrame(data[1])

    df = df[['countryiso3code','date','value']]

    df.columns = ['country','year',indicator]

    df['year'] = pd.to_numeric(df['year'], errors='coerce')

    df[indicator] = pd.to_numeric(df[indicator], errors='coerce')

    return df


# PIB PER CAPITA
df_gdp = download_worldbank("NY.GDP.PCAP.KD")

df_gdp.rename(columns={"NY.GDP.PCAP.KD":"gdp_per_capita"}, inplace=True)

df_gdp = df_gdp[df_gdp['year'] >= 2010]


# PARTICIPAÇÃO FEMININA NA FORÇA DE TRABALHO
df_female = download_worldbank("SL.TLF.CACT.FE.ZS")

df_female.rename(columns={"SL.TLF.CACT.FE.ZS":"participação_feminina_na_STEM"}, inplace=True)

df_female = df_female[df_female['year'] >= 2010]

# MERGE DOS DATASETS
df = pd.merge(df_gdp, df_female, on=['country','year'], how='inner')

df = df.dropna()

df = df[(df['gdp_per_capita'] > 0) & (df['participação_feminina_na_STEM'] > 0)]


# MÉDIA POR PAÍS
df_country = df.groupby('country').mean(numeric_only=True).reset_index()

# GRÁFICO 1 — CORRELAÇÃO
plt.figure(figsize=(10,6))

sns.regplot(
    data=df_country,
    x='participação_feminina_na_STEM',
    y='gdp_per_capita'
)

plt.title("Participação das Mulheres na STEM vs PIB per Capita")

plt.xlabel("Participação das Mulheres na STEM (%)")

plt.ylabel("PIB per Capita")

plt.show()

# GRÁFICO 2 — TENDÊNCIA GLOBAL
df_global = df.groupby('year').mean(numeric_only=True).reset_index()

plt.figure(figsize=(10,6))

plt.plot(df_global['year'], df_global['participação_feminina_na_STEM'])

plt.title("Participação das Mulheres na STEM - Tendência Global")

plt.xlabel("Ano")

plt.ylabel("Participação das Mulheres na STEM (%)")

plt.show()

# GRÁFICO 3 — MAPA MUNDIAL
fig = px.choropleth(

    df_country,

    locations="country",

    color="participação_feminina_na_STEM",

    hover_name="country",

    color_continuous_scale="viridis",

    title="Participação das Mulheres na Força de Trabalho - Mapa Mundial"

)

fig.show()

# SALVAR DADOS TRATADOS
df_country.to_csv("female_participation_vs_economy.csv", index=False)

print("Arquivo salvo com sucesso!")
```
